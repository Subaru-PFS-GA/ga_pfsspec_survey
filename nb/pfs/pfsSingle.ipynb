```python
fn, identity = repo.locate_product(PfsSingle, visit=VISIT, objId=OBJID, catId=CATID, run=RUN)
spec, _, _ = repo.load_product(product=PfsSingle, filename=fn)

fn, identity
```

```python
f, axs = plt.subplots(2, 1, figsize=(12, 4), dpi=300, sharex=True)

wave = spec.fluxTable.wavelength
flux = spec.fluxTable.flux
error = spec.fluxTable.error
mask = ~np.isnan(flux) & ~np.isinf(flux)

axs[0].plot(wave, flux, lw=0.3)
axs[0].plot(wave, error, lw=0.3)
axs[0].set_ylim(0, 1.1 * np.quantile(flux[mask], 0.95))
axs[0].grid()

axs[1].plot(0.5 * (wave[1:] + wave[:-1]), np.diff(wave), lw=0.5)
axs[1].plot(wave, np.zeros_like(wave), '.')
# axs[1].set_xlim(850, 950)
# axs[1].set_ylim(0.08, 0.081)
axs[1].grid()
```
