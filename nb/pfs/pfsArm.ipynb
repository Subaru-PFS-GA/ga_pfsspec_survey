```python
import numpy as np
import astropy.io.fits
import matplotlib.pyplot as plt
```

```python
from pfs.datamodel import PfsArm
```

```python
# Open the file with 
import astropy.io.fits
with astropy.io.fits.open(DATA_FILE) as fits:
    for hdu in fits:
        print(hdu.name)
        if isinstance(hdu, astropy.io.fits.BinTableHDU):
            print('BINTABLE', hdu.name, hdu.data.shape)
        elif isinstance(hdu, astropy.io.fits.ImageHDU):
            print('IMAGE', hdu.data.shape)
        else:
            print('--')
```

```python
with astropy.io.fits.open(DATA_FILE) as fits:
    wave = fits['WAVELENGTH'].data
    flux = fits['FLUX'].data
    sky = fits['SKY'].data
    norm = fits['NORM'].data
    covar = fits['COVAR'].data

wave.shape, flux.shape, sky.shape, norm.shape, covar.shape
```

```python
i = 5

fig, axes = plt.subplots(2, 1, figsize=(12, 6), dpi=240)

axes[0].plot(wave[i], flux[i], label='flux', lw=0.3)
# axes[0].plot(wave[i], sky[i] * norm[i], label='sky')
# axes[0].plot(wave[i], norm[i], label='norm', lw=0.3)

axes[0].set_ylim(-0.1 * np.quantile(flux[i], 0.01), 1.1 * np.quantile(flux[i], 0.99))

x = wave[i]
y = np.sqrt(covar[i, 0, :])
m = np.isfinite(y) & (y > 0)
axes[1].plot(np.where(m, x, np.nan), np.where(m, y, np.nan), label='covar', lw=0.3)
axes[1].set_ylim(0, 1.1 * np.quantile(y[m], 0.99))
```

```python
i = 5

fig, axes = plt.subplots(2, 1, figsize=(12, 6), dpi=240)

axes[0].plot(wave[i], flux[i], label='flux', lw=0.3)
axes[0].plot(wave[i], sky[i] * norm[i], label='sky')
axes[0].plot(wave[i], norm[i], label='norm', lw=0.3)

x = wave[i]
y = np.sqrt(covar[i, 0, :])
m = np.isfinite(y) & (y > 0)
axes[1].plot(np.where(m, x, np.nan), np.where(m, y, np.nan), label='covar', lw=0.3)
axes[1].set_ylim(0, 1.1 * np.quantile(y[m], 0.99))
```
